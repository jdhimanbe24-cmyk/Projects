```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import mean_absolute_error
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense, Dropout
from tensorflow.keras.callbacks import EarlyStopping, ReduceLROnPlateau
from tensorflow.keras.optimizers import Adam
```

```python
cols = ['engine_id', 'cycle'] + \
       [f'op_setting_{i}' for i in range(1, 4)] + \
       [f'sensor_{i}' for i in range(1, 22)]

train = pd.read_csv('/content/train_FD001.txt', sep=r'\s+', header=None)
train = train.iloc[:, :26]
train.columns = cols
```

```python
train['RUL'] = train.groupby('engine_id')['cycle'].transform('max') - train['cycle']
train['RUL'] = train['RUL'].clip(upper=125)
```

```python
important_sensors = ['sensor_2', 'sensor_3', 'sensor_4', 'sensor_7', 'sensor_11']

def add_features(df):
    df = df.copy()

    for col in important_sensors:
        df[col + '_diff'] = df[col].diff()
        df[col + '_roll'] = df[col].shift(1).rolling(5).mean()

    df.bfill(inplace=True)
    df.ffill(inplace=True)
    return df

train = add_features(train)
```

```python
feature_cols = train.columns.drop(['engine_id', 'cycle', 'RUL'])

scaler = StandardScaler()
train[feature_cols] = scaler.fit_transform(train[feature_cols])
```

```python
def create_sequences(df, seq_length=50, stride=1):
    X, y = [], []

    for eng in df['engine_id'].unique():
        temp = df[df['engine_id'] == eng].sort_values('cycle')

        values = temp[feature_cols].values
        rul = temp['RUL'].values

        for i in range(0, len(temp) - seq_length):
            X.append(values[i:i+seq_length])
            y.append(rul[i + seq_length - 1])

    return np.array(X), np.array(y)

X_train, y_train = create_sequences(train)
print(X_train.shape, y_train.shape)
```

```python
model = Sequential([
    LSTM(64, return_sequences=True, input_shape=(X_train.shape[1], X_train.shape[2])),
    Dropout(0.4),

    LSTM(32),
    Dropout(0.4),

    Dense(32, activation='relu'),
    Dense(1)
])

model.compile(
    optimizer=Adam(learning_rate=0.0005),
    loss='mse',
    metrics=['mae']
)
```

```python
early_stop = EarlyStopping(
    monitor='val_mae',
    patience=8,
    restore_best_weights=True
)

lr_reduce = ReduceLROnPlateau(
    monitor='val_mae',
    factor=0.5,
    patience=3,
    min_lr=1e-6
)

history = model.fit(
    X_train, y_train,
    epochs=50,
    batch_size=32,
    validation_split=0.2,
    callbacks=[early_stop, lr_reduce],
    verbose=1
)
```

```python
test = pd.read_csv('/content/test_FD001.txt', sep=r'\s+', header=None)
test = test.iloc[:, :26]
test.columns = cols
```

```python
rul = pd.read_csv('/content/RUL_FD001.txt', header=None)
rul.columns = ['RUL']

max_cycle = test.groupby('engine_id')['cycle'].max().reset_index()
max_cycle.columns = ['engine_id', 'max_cycle']

max_cycle['RUL'] = rul['RUL']

test = test.merge(max_cycle, on='engine_id', how='left')
test['RUL'] = test['RUL'] + (test['max_cycle'] - test['cycle'])
```

```python
test = add_features(test)
```

```python
assert list(train[feature_cols].columns) == list(test[feature_cols].columns)
```

```python
test[feature_cols] = scaler.transform(test[feature_cols])
```

```python
X_test, y_test = create_sequences(test)
print(X_test.shape, y_test.shape)
```

```python
y_pred = model.predict(X_test).flatten()

mae = mean_absolute_error(y_test, y_pred)
print("TEST MAE:", mae)
```
