```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense, Dropout
from tensorflow.keras.callbacks import EarlyStopping, ReduceLROnPlateau
from tensorflow.keras.optimizers import Adam
```

```python
cols = ['engine_id', 'cycle'] + \
       [f'op_setting_{i}' for i in range(1, 4)] + \
       [f'sensor_{i}' for i in range(1, 22)]

df = pd.read_csv('/content/train_FD001.txt', sep=r'\s+', header=None)
df = df.iloc[:, :26]
df.columns = cols
```

```python
df['RUL'] = df.groupby('engine_id')['cycle'].transform('max') - df['cycle']

# scale RUL (IMPORTANT)
df['RUL'] = df['RUL'] / 125.0
```

```python
important_sensors = ['sensor_2', 'sensor_3', 'sensor_4', 'sensor_7', 'sensor_11']

def add_features(df):
    for col in important_sensors:
        df[col + '_diff'] = df[col].diff()
        df[col + '_roll'] = df[col].rolling(5).mean()

    df.bfill(inplace=True)
    df.ffill(inplace=True)
    return df

df = add_features(df)
```

```python
feature_cols = df.columns.drop(['engine_id', 'cycle', 'RUL'])

scaler = StandardScaler()
df[feature_cols] = scaler.fit_transform(df[feature_cols])
```

```python
def create_sequences(df, seq_length=50, stride=2):
    X, y = [], []

    for eng in df['engine_id'].unique():
        temp = df[df['engine_id'] == eng].sort_values('cycle')

        values = temp[feature_cols].values
        rul = temp['RUL'].values

        for i in range(0, len(temp) - seq_length, stride):
            X.append(values[i:i+seq_length])
            y.append(rul[i+seq_length])

    return np.array(X), np.array(y)

X, y = create_sequences(df)
print(X.shape, y.shape)
```

```python
model = Sequential()

model.add(LSTM(48, return_sequences=True,
               input_shape=(X.shape[1], X.shape[2])))
model.add(Dropout(0.45))

model.add(LSTM(24))
model.add(Dropout(0.45))

model.add(Dense(32, activation='relu'))
model.add(Dense(1))
```

```python
model.compile(
    optimizer=Adam(learning_rate=0.0005),
    loss='mse',
    metrics=['mae']
)
```

```python
early_stop = EarlyStopping(
    monitor='val_mae',
    patience=8,
    restore_best_weights=True
)

lr_reduce = ReduceLROnPlateau(
    monitor='val_mae',
    factor=0.5,
    patience=3,
    min_lr=1e-6
)

history = model.fit(
    X, y,
    epochs=50,
    batch_size=32,
    validation_split=0.2,
    callbacks=[early_stop, lr_reduce],
    verbose=1
)
```
